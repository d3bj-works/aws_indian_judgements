# Canonical Statute Standardization & Enactment Tagging (Central vs State)
**Corpus Scope**: 38,235 Supreme Court Judgments (1950–2026)  
**Source Asset**: [`notebooks/reports/assets/acts_and_sections.csv`](file:///home/duttadev/projects/aws_indian_judgements/notebooks/reports/assets/acts_and_sections.csv)  
**Primary Objective**: Fuzzy-match variant enactment names into standardized canonical URIs and distinguish Central Enactments from State Enactments.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

CSV_PATH = 'reports/assets/acts_and_sections.csv'
df = pd.read_csv(CSV_PATH)
print(f'Dataset loaded successfully: {len(df):,} Act-Section rows.')
print(f'Columns: {list(df.columns)}')
```

## 1. Executive Summary & Resolution Impact
By mapping acronyms (`IPC`, `CrPC`, `CPC`, `IEA`, `NDPS`, `BNS`, `BNSS`, `BSA`) and string variants into canonical titles and legal URIs (`urn:lex:in:act:...`), we resolve entity fragmentation across the corpus.

```python
raw_count = df['statute'].nunique()
canonical_count = df['canonical_title'].nunique()
reduction_pct = round((1 - (canonical_count / raw_count)) * 100, 2)
print(f'Raw Variant Enactment Strings: {raw_count:,}')
print(f'Consolidated Canonical Titles: {canonical_count:,}')
print(f'Entity Resolution Impact: {reduction_pct}% reduction in statutory name fragmentation.')
```
